```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
```

```python
ds1 = xr.open_dataset("daily_maps/Outliers/SCFG/usa_Alaska_19920612_gac.nc")
```

```python
ds2 = xr.open_dataset("daily_maps/Outliers/SCFG/usa_Alaska_19920612_lac.nc")
```

```python
ds3 = xr.open_dataset("daily_maps/Outliers/SCFG/usa_Alaska_19920622_gac.nc")
```

```python
ds4 = xr.open_dataset("daily_maps/Outliers/SCFG/usa_Alaska_19920622_lac.nc")
```

```python
ds5 = xr.open_dataset("daily_maps/Outliers/SCFV/usa_Alaska_19920612gac.nc")
```

```python
ds6 = xr.open_dataset("daily_maps/Outliers/SCFV/usa_Alaska_19920612lac.nc")
```

SCFG

```python
plot_data = ds1["scfg"].sel(
    lat=slice(66.165, 68.155),
    lon=slice(-153.365, -151.375)
)

# Plot
plt.figure(figsize=(8, 6))

plot_data.plot()

plt.title("SCFG GAC Alaska - 1992-06-12")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.tight_layout()

# Save figure
plt.savefig(
    "plots/Anomalies/GAC_Alaska_1992-06-12.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plot_data = ds2["scfg"].sel(
    lat=slice(66.165, 68.155),
    lon=slice(-153.365, -151.375)
)

# Plot
plt.figure(figsize=(8, 6))

plot_data.plot()

plt.title("SCFG LAC Alaska - 1992-06-12")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.tight_layout()

# Save figure
plt.savefig(
    "plots/Anomalies/LAC_Alaska_1992-06-12.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Select the same 2° × 2° Alaska region from both datasets
lac = ds1["scfg"].sel(
    lat=slice(66.165, 68.155),
    lon=slice(-153.365, -151.375)
)

gac = ds2["scfg"].sel(
    lat=slice(66.165, 68.155),
    lon=slice(-153.365, -151.375)
)

# Calculate difference: LAC - GAC
diff = lac - gac

# Plot difference map
fig, ax = plt.subplots(figsize=(8, 6))

diff.plot(
    ax=ax,
    cmap="RdBu_r",
    vmin=-100,
    vmax=100,
    cbar_kwargs={"label": "SCFV difference (%)"}
)

ax.set_title("SCFG Alaska SCFV difference\n(LAC - GAC, 1992-06-12)")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")

plt.tight_layout()

# Save figure
plt.savefig(
    "plots/Anomalies/LAC_GAC_Alaska_difference_1992-06-12.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
