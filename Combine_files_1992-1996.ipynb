```python
import os
import pandas as pd

input_folder = "csv/SCFG_2x2_degee"
output_folder = "csv_by_region"

os.makedirs(output_folder, exist_ok=True)

all_data = []

# Read all CSV files in all subfolders
for root, folders, files in os.walk(input_folder):
    for file in files:
        if file.endswith(".csv"):
            path = os.path.join(root, file)

            df = pd.read_csv(path)
            all_data.append(df)

# Combine all CSV files
data = pd.concat(all_data, ignore_index=True)

# Create separate file for each region
for region, group in data.groupby("region"):

    output_file = os.path.join(
        output_folder,
        f"{region}.csv"
    )

    group.to_csv(
        output_file,
        index=False
    )

    print("Saved:", output_file)
```
